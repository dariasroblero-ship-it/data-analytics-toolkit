## 1. Bring in our tools
`pandas` is the library for working with tables of data. We nickname it `pd`. `matplotlib` is for charts.

```python
import pandas as pd              # pandas: tables of data. 'pd' is the standard nickname.
import matplotlib.pyplot as plt   # matplotlib.pyplot: charts. 'plt' is the standard nickname.
```

## 2. Load the data
This is a real dataset: 244 restaurant bills, each with the total bill, the tip, the day, and more. `read_csv` loads the file into a table we call a **DataFrame**, stored in `df`.

```python
df = pd.read_csv('../data/tips.csv')   # read the CSV file into a DataFrame (a table) called df
df.head()                              # .head() shows just the first 5 rows so we can eyeball it
```

`head()` shows the first 5 rows. Each row is one table (one bill). The columns are what we know about it.

## 3. How much data do we have?

```python
df.shape   # gives (number_of_rows, number_of_columns) -> here (244, 6)
```

244 rows, 6 columns. 244 real restaurant visits.

## 4. Ask your first question: what is the average tip?

```python
df['tip'].mean()   # df['tip'] grabs the tip column; .mean() averages all its values
```

`df['tip']` grabs the tip column, and `.mean()` averages it. You just computed a statistic from real data.

## 5. Does the tip grow with the bill?
Let's *see* it. Each dot is one visit: the bill on the x-axis, the tip on the y-axis.

```python
# scatter plot: one dot per visit. x = total_bill, y = tip.
plt.scatter(df['total_bill'], df['tip'],   # the x values, then the y values
            alpha=0.5,                     # alpha = dot transparency, so overlapping dots show
            color='#00053E')               # ETSU navy
plt.xlabel('total bill ($)')               # label the x-axis
plt.ylabel('tip ($)')                      # label the y-axis
plt.title('Do bigger bills bring bigger tips?')  # chart title
plt.show()                                 # render the chart
```

The dots drift upward to the right: bigger bills tend to bring bigger tips. You just found a relationship in data.

## 6. Which day earns the best tips?
`groupby('day')` splits the rows by day, then we average the tip in each group.

```python
# split rows by day, then average the tip within each day
avg_tip_by_day = df.groupby('day')['tip'].mean()   # group by 'day', take mean of 'tip' per group
print(avg_tip_by_day)                              # show the number for each day

# turn that result into a bar chart
avg_tip_by_day.plot(kind='bar',          # make a bar chart from the grouped result
                    color='#FFC72C',     # ETSU gold bars
                    edgecolor='#00053E') # navy outline on each bar
plt.ylabel('average tip ($)')            # label the y-axis
plt.title('Average tip by day')          # chart title
plt.show()                               # render the chart
```

## That's it, you just did data analysis.

In ten minutes you loaded real data, computed a statistic, found a relationship, and compared groups. That is the whole job in miniature.
